```python
df.quality.value_counts()
df.quality.value_counts().plot(kind = 'bar', color = 'g')
plt.xlabel('Wine Quality')
plt.ylabel('Count')
plt.show()
```

```python
df.density.value_counts()
df.density.value_counts().plot(kind = 'bar', color = 'g')
plt.xlabel('Density')
plt.ylabel('Count')
plt.show()
```

```python
for column in df.columns:
    sns.histplot(df[column], kde = True)
```

```python
sns.histplot(df['alcohol'], color = 'purple')
```

```python
sns.histplot(df['pH'], color = 'blue')
```

```python
quality_counts = df['quality'].value_counts().sort_index()
```

```python
plt.figure(figsize=(8, 8))
plt.pie(
    quality_counts,
    labels=quality_counts.index,
    autopct='%1.1f%%',
    startangle=180
)
plt.title("Distribution of White Wine Quality")
plt.show()
```
