```python
import numpy as np
from sklearn.metrics import confusion_matrix, classification_report
import seaborn as sns
import matplotlib.pyplot as plt
```

 FOR 2 VARIABLES ; DOG and NOT DOG

```python
actual = np.array(
    ['Dog', 'Dog', 'Dog', 'Not Dog', 'Dog', 'Not Dog', 'Dog', 'Dog', 'Not Dog', 'Not Dog'])
predicted = np.array(
    ['Dog', 'Not Dog', 'Dog', 'Not Dog', 'Dog', 'Dog', 'Dog', 'Dog', 'Not Dog', 'Not Dog'])
```

```python
actual 
```

```python
predicted
```

```python
cm = confusion_matrix(actual, predicted)
```

```python
cm
```

```python
sns.heatmap(cm,
           annot = True,
           fmt = 'g',
           xticklabels = ['Dog', 'Not Dog'],
           yticklabels = ['Dog', 'Not Dog'])
plt.ylabel('Actual', fontsize = 12)
plt.title('Confusion Matrix', fontsize = 15, pad = 20)
plt.gca().xaxis.set_label_position('top')
plt.xlabel('Predicted', fontsize = 12)
plt.gca().xaxis.tick_top()

plt.gca().figure.subplots_adjust(bottom = 0.2)
plt.gca().figure.text(0.5, 0.05, 'Prediction', ha = 'center', fontsize = 13)

plt.show()
```

```python
print(classification_report(actual, predicted))
```

FOR MULTI-CLASS ; CAT, DOG AND HORSE

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay, classification_report
```

```python
y_true = ['Cat'] * 10 + ['Dog'] * 12 + ['Horse'] * 10
y_pred = ['Cat'] * 8 + ['Dog'] + ['Horse'] + ['Cat'] * 2 + ['Dog'] * 10 + ['Horse'] * 8 + ['Dog'] * 2
classes = ['Cat', 'Dog', 'Horse']
```

```python
cm = confusion_matrix(y_true, y_pred, labels = classes)
```

```python
cm
```

```python
disp = ConfusionMatrixDisplay(confusion_matrix = cm, display_labels = classes)
disp.plot(cmap = plt.cm.Blues)
plt.title('Confusion Matrix', fontsize = 12, pad = 20)
plt.xlabel('Predicted', fontsize = 8)
plt.ylabel('Actual', fontsize = 8)
plt.gca().xaxis.set_label_position('top')
plt.gca().xaxis.tick_top()
plt.gca().figure.subplots_adjust(bottom = 0.2)
plt.gca().figure.text(0.5, 0.05, 'Predicted', ha = 'center', fontsize = 12)

plt.show()
```
